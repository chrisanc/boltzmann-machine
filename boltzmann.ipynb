## **Introducción**

Una máquina de Boltzmann es un tipo especifico de red neuronal estocástica cuya estructura incluye unidades de procesamiento interconectadas que operan de manera probabilistica. A diferencia de otras redes neuronales, las máquinas de Boltzmann son especialmente útiles para modelar y simular sistemas complejos y dinámicos debido a que pueden capturar la variabilidad e incertidumbre inherentes en muchos problemas del mundo real [1]. Una de sus características principales es que se considera una red neuronal *recurrente*, las cuales se distinguen por su "memoria" ya que toman información de entradas anteriores para influir en la entrada y salida actual [2].

**¿Cómo puede una red neuronal estocástica aprender una distribución de probabilidad y cuáles son las dificultades que limitan su entrenamiento?** Para poder responder a esta pregunta, es necesario primero entender los conceptos básicos que nos llevan a cuestionarla.
Un *proceso estocástico* es todo proceso que describe la evolución de un sistema a lo largo del tiempo donde el azar tiene una intervención. A diferencia de los modelos deterministas, donde el resultado está totalmente definido por las condiciones iniciales, en estos procesos hay incertidumbre en los resultados [3].

De este modo, podemos definir a las *redes neuronales estocásticas* como modelos caracterizados por las variaciones aleatorias ya sea en sus neuronas o en sus pesos. Estas resultan especialmente útiles para resolver problemas de optimización ya que, gracias a estas fluctuaciones, podemos crear algoritmos que eviten los minimos locales además de modelar y simular sistemas complejos.

Su principal limitación son los tiempos de entrenamiento, ya que es extremadamente lento y computacionalmente caro debido a la necesidad de calcular el equilibrio termico mediante métodos de muestreo avanzados.

## **Antecedentes**

En 1982, John Hopfield introdujo la *Red de Hopfield*, una red neuronal recurrente de unidades binarias con conexiones simétricas que funcionaba como un sistema dinámico con memoria asociativa. Su funcionamiento es simple: dada una configuración inicial, la red actualiza asincronicamente el estado de sus nodos hasta lograr la convergencia a un estado estable. Este modelo, a pesar de haber revolucionado el área de la inteligencia artificial, su dinámica descendente provocaba que la red se quedará bloqueada en configuraciones no óptimas sin poder alcanzar un mínimo global [4].

Para resolver las limitaciones en sistemas energéticos, se desarrolló la técnica de *Recocido Simulado*, el cual introduce ruido estocástico controlado en las reglas de transición entre estados para que el sistema adquiera la capacidad de realizar saltos probabilisticos hacia estados de mayor energía. Cuando se obtienen temperaturas elevadas, el sistema explora el espacio disponible de estados; a medida que la temperatura disminuye, su probabilidad de visitar estados desfavorables también [5].

En 1985 se consolidó la *máquina clásica de Boltzmann*, unificando la estructura de la *red de Hopfield* y la técnica del *Recocido Simulado*. Entre sus tantas innovaciones y contribuciones teoricas están [2]:

- Se dividieron las neuronas en unidades visibles (las cuales reciben la entrada/salida de datos) y unidades ocultas (las cuales capturan correlaciones complejas).

- Se reemplazó la función de activación determinista por reglas de transición probabilistica.

- Se demostró que, al correr la red libremente a una temperatura fija durante el tiempo suficiente, la probabilidad de observar cualquier estado global converge a la distribución estocástica de Boltzmann.

A pesar de tener un sustento y solidez matemática, este modelo resulta computacionalmente inviable para aplicaciones prácticas debido al intenso costo computacional que representa utilizarlo. Se propusieron algunos modelos derivados que superan este obstaculo, como la *Máquina de Boltzmann Restringida* o la *Restricción Bipartita* [6].

## **Arquitectura**

La arquitectura de una máquina de Boltzmann se puede describir con el siguiente diagrama:

![classic](./assets/clasica.png)

En donde:

- **Unidades visibles:** Representan las variables observables del entorno (datos de entrada) [7].

- **Unidades ocultas:** Actúan como detectores de características no lineales que capturan correlaciones complejas entre variables visibles [7].

- **Pesos (W):** Representa el peso de la conexión simétrica entre la unidad visible y la unidad oculta, es decir, que tan importante será esa entrada en la toma de decisiones [7].

- **Sesgos:** Cada unidad visible/oculta tiene un valor de sesgo, el cual accion a cada unidad para que ejecute una acción y nunca quede sin hacer nada (muerta) [7].

Cada una de estas unidades tiene dos estados posibles: 0 o 1. Es decir, sus estados son totalmente binarios.

Como podemos notar en esta arquitectura, la clave es que cada una de las neuronas está conectada entre si, incluso las de la misma capa, lo que hace que cada unidad sea dependiente de otras las cuales provienen de multiples capas.

En cambio, la arquitectura de la máquina de Boltzmann restrigida es:

![restricted](./assets/restringida.png)

Podemos notar que la principal diferencia entre ambas es que esta (RBM) prohibe explicitamente las conexiones internas dentro de la misma capa. Esto garantiza que, dado el estado de la capa visible todas las unidades ocultas y visibles son condicionalmente independientes entre si por lo que se permite realizar inferencia probabilística [8].

## **Comportamiento probabilístico**

El estado de las unidades en arquitecturas deterministas (como la red de Hopfield o perceptrón) se rige por la función escalón o umbral, la cual indica que si una neurona se activa basandose en el valor de la suma ponderada. El uso de esta función es bastante simple y entendible, pero tiene un problema en modelos basados en energía: no permite transicionar a estados de mayor energía, lo que ocasiona que pueda quedar atrapado en mínimos locales. Además, elimina el factor estocástico el cual es necesario para permitir al modelo atravesar barreras energéticas, escapar de mínimos locales o explorar el espacio de estados a su conveniencia [].

La expresión utilizada es la siguiente:

$P(s_i = 1) = \frac{1}{1 + \mathrm{e}^{\frac{-\Delta E_i}{T}}}$

Donde:

- $P(s_i = 1)$: Representa la probabilidad de que la neurona cambie o se mantenga en el estado activo, dado el estado del resto de las neuronas presentes en la red.

- $s_i$: Variable estocástica binaria que representa su estado actual (0 o 1).

- $\Delta E_i$: Representa la diferencia de energia o la entrada neta, siendo la variación global que experimenta la red cuando alguna neurona se activa.

- $T$: Representa el nivel de ruido estocástico o fluctuación térmica del sistema. En caso de que $T = 0$, esta actuará como una función escalón convencional, conforme aumenta su valor el ruido que introduce también lo hace y cuando disminuye el ruido que introduce también lo hace.

- $e$: Constante de Euler, la cual es la base de los logaritmos naturales.

- $\frac{1}{1 + \mathrm{e}^{\frac{-\Delta E_i}{T}}}$: Representa la función logistica  la cual toma cualquier valor de entrada y lo mapea a un rango continuo entre 0 y 1, lo que nos permite interpretarlo como una probabilidad.

Al realizar dos ejecuciones, el sistema produce dos secuencias de estados diferentes debido a la naturaleza estocástica inherente al modelo y a su mecanismo de transiciones probabilisticas. El estado global de la red ($\Delta E_i$) se actualiza paso a paso mediante el *Muestreo de Gibbs*, realizando un proceso de "cadena" donde cada estado $x_{i + 1}$ se muestrea probabilisticamente con todos sus estados anteriores.

## **Función de energía**

Para definir estados binarios como energia, se utiliza la siguiente ecuación:

$E(s) = - \frac{1}{2} \sum_{i} \sum_{j} w_{ij} s_i s_j - \sum_{i} b_i s_i$

Esta ecuación define como calcular la energía global del estado ($E(s)$) basado en los siguientes parámetros:

- $E(s)$: Valor escalar que mide el nivel de conflicto o desacuerdo en la red para una configuración particular de neuronas $s$. A menor valor de energía, menor será la probabilidad de que el sistema visite o genere dicho estado.

- $w_{ij}$: Representa la intensidad de la conexión e interacción entre la neurona $i$ y la neurona $j$. Entre mayor sea el valor, la red premiará la activación simultanea de ambas neuronas por lo que la energía disminuirá.

- $s_i, s_j$: Variables estocásticas y binarias de las neuronas $i$ y $j$. El valor de su producto es máximo 1.

- Factor $\frac{1}{2}$: Al realizar una doble suma sobre todos los indices $i$ y $j$, cada par de neuronas $(i, j)$ y $(j, i)$ se contabiliza dos veces, por lo que se debe corregir para que solo contributa una sola vez a la energía total del sistema.

- $b_i$: Parámetro que determina la preferencia inherente de la neurona $i$ a estar activa, independientemente del estado actual de las demás neuronas.

- $s_i$ (Segundo termino): Representa el estado de la neurona $i$, multiplicando al sesgo $b_i$, lo que garantiza que el sesgo solo altere la energía global cuando la neurona esté encendida. 

Ahora, *¿por qué el signo negativo?* Este garantiza que las configuraciones deseables reduzcan la energía global y, por ende, tengan mayor probabilidad de ocurrencia.

El comportamiento probabilistico general de estas máquinas se basa en la conexión de la energía, estabilidad y probabilidad.

La relación entre la energía y la probabilidad en las máquinas de Boltzmann está formalizada por la distribución de Gibbs-Boltzmann, según la cual la probabilidad conjunta de que el sistema se encuentre en un estado global determinado $s$ disminuye de manera exponencial a medida que aumenta su función de energía $E(s)$. Las configuraciones con menor energía adquieren una mayor probabilidad de ocurrencia al representar estados estables que satisfacen óptimamente las restricciones de la red, mientras que los estados de alta energía son penalizados y resultan poco probables

## **Distribución de Boltzmann**

La probabilidad normalizada de un estado se define por:

$P(s) = \frac{\mathrm{e}^{\frac{-E(s)}{T}}}{Z}$

Esto debido a que las configuraciones de menor energía resultan en una mayor probabilidad.

Su función de partición es:

$Z = \sum_{s} \mathrm{e}^{\frac{-E(s)}{T}}$

La función de partición $Z$ constituye la constante de normalización de la distribución de probabilidad de Gibbs-Boltzmann, encargada de transformar los factores energéticos unificados $e^{-E(s)/T}$ en valores de probabilidad formalmente válidos.
Su cálculo exige considerar la totalidad de los estados posibles del sistema porque, de acuerdo con los fundamentos de la probabilidad, la suma de las probabilidades asignadas a cada una de las configuraciones dentro del espacio de estados debe ser exactamente igual a uno. Si $Z$ omitiera tan solo una configuración global, la constante de normalización resultante sería inferior a la real, lo que provocaría que la suma total de las probabilidades superara la unidad y desvalidaría el modelo probabilístico. Cabe señalar además que la convención matemática asigna un signo negativo a la energía en el exponente ($P(s) = \frac{e^{-E(s)/T}}{Z}$), garantizando que los estados de menor energía posean probabilidades de ocurrencia más elevadas.

La severa dificultad computacional para evaluar $Z$ radica en el crecimiento exponencial del espacio de estados a medida que aumenta el número de neuronas en la red. En un sistema compuesto por $N$ unidades binarias (visibles y ocultas), cada neurona puede adoptar dos estados independientes ($0$ o $1$), lo que genera un total de $2^N$ configuraciones globales posibles. Como la suma requerida para evaluar $Z$ debe iterar sobre este universo completo de $2^N$ estados, el número de operaciones matemáticas requeridas crece con una complejidad temporal de $\mathcal{O}(2^N)$. Mientras que para redes de dimensiones reducidas el cálculo es directo, al escalar a tamaños típicos en aplicaciones reales (por ejemplo, con 100 neuronas), el número de estados a evaluar alcanza $2^{100} \approx 1.26 \times 10^{30}$, cifra que sobrepasa la capacidad de cómputo de cualquier supercomputadora.

Esta explosión combinatoria convierte la evaluación exacta de la función de partición, del gradiente de verosimilitud y de las probabilidades marginales en un problema computacionalmente intratable. Por este motivo, las implementaciones prácticas de las Máquinas de Boltzmann y de las Máquinas de Boltzmann Restringidas (RBM) evitan calcular $Z$ de forma directa durante el entrenamiento. En su lugar, emplean algoritmos de aproximación estocástica basados en Cadenas de Markov y Muestreo de Monte Carlo (MCMC) (como el Muestreo de Gibbs y la Divergencia Contrastiva), los cuales estiman las estadísticas requeridas mediante muestras representativas sin necesidad de sumar exhaustivamente sobre todo el espacio de estados.
